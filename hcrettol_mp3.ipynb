```python
# agreement stats for each field
stats_rows = []
for field in ['HasRecovered', 'WhyRecovered', 'ActivityPattern', 'Currentstatus', 'RecentThemes']:
    f = log[log['Field'] == field]
    compared = f[f['PreDiscussAgreementstatus(Yes/No)'] != 'NA']   # rows where rpatel92 gave a value

    if len(compared) == 0:
        stats_rows.append({'Field': field, 'PercentAgree_Pre': 'NA', 'PercentAgree_Post': 'NA',
                           'CohenKappa_Pre': 'NA', 'CohenKappa_Post': 'NA',
                           'NumDisagreements': 'NA', 'NumResolved': 'NA',
                           'MeanResolutionTime_min': 'NA',
                           'Notes': 'Not in rpatel92s stats file so nothing to compare'})
        continue

    agree_pre = (compared['PreDiscussAgreementstatus(Yes/No)'] == 'Yes').mean()
    agree_post = 1.0   # all disagreements got resolved
    num_dis = (compared['PreDiscussAgreementstatus(Yes/No)'] == 'No').sum()

    # Cohen's kappa from the slides: (Pa - Ps) / (1 - Ps), Ps = 1 / number of labels
    # ActivityPattern has 7 possible labels in the coding guide
    ps = 1 / 7
    kappa_pre = round((agree_pre - ps) / (1 - ps), 2)
    kappa_post = round((agree_post - ps) / (1 - ps), 2)

    stats_rows.append({'Field': field,
                       'PercentAgree_Pre': round(agree_pre * 100, 1),
                       'PercentAgree_Post': round(agree_post * 100, 1),
                       'CohenKappa_Pre': kappa_pre, 'CohenKappa_Post': kappa_post,
                       'NumDisagreements': num_dis, 'NumResolved': num_dis,
                       'MeanResolutionTime_min': 'NA',
                       'Notes': 'rpatel92 called almost everything cyclical but most plots looked declining'})

agreement = pd.DataFrame(stats_rows)
agreement.to_csv(netid + '_agreement_stats.csv', sep=';', index=False)

# reflection
reflection = """# Validation Reflection - hcrettol

My partner didn't have his MiniProject2 posted, so the TA gave me rpatel92's to validate instead. His stats file only had ActivityPattern, so that was the only field I could actually compare. For the other four fields I just put my own answers based on his commit data and plots.

ActivityPattern had the most disagreement. rpatel92 called 9 of his 10 projects cyclical, but most of them looked declining to me, with a big peak early and less activity after. The coding guide says cyclical means activity repeats at fixed intervals, and most of these just had spikes at random times. pyprocar was the hardest one since it has a lot of spikes, but they weren't on a regular schedule so I said irregular.

Before the long gaps, a lot of the projects were only doing small things like version bumps and dependency fixes. When they came back it was usually either the same main developer returning or new people fixing stuff that broke with newer libraries, like numpy for Calamari and scipy for batchgenerators. Most of the projects did recover and 7 of the 10 are still active, so a gap didn't usually mean the project was dead. The main thing I had to watch out for was fake recoveries. football had commits after its gap, but they were from random people and dependabot instead of the actual maintainers, so I didn't count it.
"""
with open(netid + '_validation_reflection.md', 'w') as f:
    f.write(reflection)

agreement
```
